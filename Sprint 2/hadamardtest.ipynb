# 1-Qubit Hadamard Gate Test

```python
import qiskit
```

```python
import qiskit_aer
import numpy as np
import matplotlib.pyplot as plt
```

```python
from qiskit import QuantumCircuit, transpile, ClassicalRegister, QuantumRegister
from qiskit_aer import AerSimulator

backend = AerSimulator() # local simulator

qc = QuantumCircuit(1) # 1 qubit circuit
qc.h(0) #applying hadamard gate on 1 qubit

qc.draw(output="mpl")
```

In this next block of code, I will run the circuit for 1000 runs and measure the value. Then I will convert them to percentages and print the results.

```python
qc.measure_all() # measures all of the qubits (in this case it is )

circ = transpile(qc, backend) #optimizing

runs = 1000
result = backend.run(qc, shots = runs).result()
count = result.get_counts(qc)
print(count)
```

```python
zero = count['0']
one = count['1']
zp = (zero/1000)*100
op = (one/1000)*100
print(f'{zp}% zeroes and {op}% ones')
```

As we can see, we have close to 50% for each. It isn't exactly 50% mainly because of noise, etc.
